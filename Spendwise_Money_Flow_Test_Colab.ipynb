```python
import base64
import io
import zipfile
from pathlib import Path

APP_DIR = Path('/content/spendwise_test_app')
APP_DIR.mkdir(parents=True, exist_ok=True)
EMBEDDED_APP = """"""
with zipfile.ZipFile(io.BytesIO(base64.b64decode(EMBEDDED_APP))) as archive:
    root = APP_DIR.resolve()
    for member in archive.infolist():
        target = (APP_DIR / member.filename).resolve()
        if not target.is_relative_to(root):
            raise ValueError('Unsafe archive path')
    archive.extractall(APP_DIR)
print('Spendwise test files extracted to', APP_DIR)
```

## Public website tunnel

Run the next cell to create a temporary public HTTPS address. The address works worldwide while this Colab runtime remains active.


```python
import os
import queue
import re
import subprocess
import threading
import time
import urllib.request
from pathlib import Path

cloudflared = Path('/content/cloudflared')
if not cloudflared.exists():
    print('Downloading Cloudflare Tunnel...')
    urllib.request.urlretrieve(
        'https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64',
        cloudflared,
    )
    cloudflared.chmod(0o755)

if 'spendwise_tunnel_process' in globals() and spendwise_tunnel_process.poll() is None:
    print('Public website:', spendwise_public_url)
else:
    spendwise_tunnel_process = subprocess.Popen(
        [str(cloudflared), 'tunnel', '--url', f'http://127.0.0.1:{spendwise_test_server.server_port}', '--no-autoupdate'],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
    )
    tunnel_lines = queue.Queue()
    def read_tunnel_output():
        for line in spendwise_tunnel_process.stdout:
            tunnel_lines.put(line)
    threading.Thread(target=read_tunnel_output, daemon=True).start()

    spendwise_public_url = None
    deadline = time.time() + 30
    while time.time() < deadline and not spendwise_public_url:
        try:
            line = tunnel_lines.get(timeout=1)
        except queue.Empty:
            continue
        match = re.search(r'https://[a-z0-9-]+\.trycloudflare\.com', line)
        if match:
            spendwise_public_url = match.group(0)

    if not spendwise_public_url:
        spendwise_tunnel_process.terminate()
        raise RuntimeError('Cloudflare could not create the public URL. Run this cell again.')
    print('Public website:', spendwise_public_url)
    print('Keep this Colab runtime running to keep the website online.')
```

```python
import importlib.util
import json
import os
import urllib.error
import urllib.request
from pathlib import Path
from threading import Thread

APP_DIR = Path('/content/spendwise_test_app')
if not (APP_DIR / 'server.py').exists():
    raise RuntimeError('Spendwise files are missing. Run Cell 1 first.')

if 'spendwise_test_server' not in globals():
    drive_root = Path('/content/drive/MyDrive')
    if drive_root.exists():
        data_dir = drive_root / 'SpendwiseData'
        data_dir.mkdir(parents=True, exist_ok=True)
        os.environ['SPENDWISE_DB'] = str(data_dir / 'spendwise_test.sqlite3')
    else:
        os.environ['SPENDWISE_DB'] = '/content/spendwise_test.sqlite3'

    spec = importlib.util.spec_from_file_location('spendwise_recovery_module', APP_DIR / 'server.py')
    spendwise_module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(spendwise_module)
    spendwise_test_server = spendwise_module.create_server(0)
    Thread(target=spendwise_test_server.serve_forever, daemon=True).start()

base = f'http://127.0.0.1:{spendwise_test_server.server_port}'

def request_status(path):
    try:
        with urllib.request.urlopen(base + path) as response:
            return response.status
    except urllib.error.HTTPError as error:
        return error.code

for path in ('/welcome.html', '/index.html', '/css/styles.css', '/js/app.js', '/js/api.js', '/js/charts.js', '/js/transactions.js'):
    assert request_status(path) == 200, path
assert request_status('/api/me') == 401
assert Path(os.environ['SPENDWISE_DB']).exists()

required_files = (
    'frontend/index.html', 'frontend/welcome.html', 'frontend/css/styles.css',
    'frontend/js/app.js', 'frontend/js/api.js', 'frontend/js/charts.js', 'frontend/js/transactions.js',
    'api/routes.py', 'api/auth.py', 'api/config.py', 'api/server.py',
    'database/storage.py', 'server.py', 'README.md', 'PRESENTATION_SCRIPT.md'
)
for name in required_files:
    source = (APP_DIR / name).read_text(encoding='utf-8')
    assert not any(marker in source for marker in ('<<<<<<<', '=======', '>>>>>>>')), name

app_source = (APP_DIR / 'frontend/js/app.js').read_text(encoding='utf-8')
for required in ('function dashboard()', 'renderTransactionsPage(', 'categoryChart(', 'recentMoneyRows()'):
    assert required in app_source, required

print('PASS: separate transaction, chart, API, database, and presentation files are included.')
```
